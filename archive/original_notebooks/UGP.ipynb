```python
import pandas as pd
import matplotlib.pyplot as plt

sheets = pd.read_excel("CPI_GR.xlsx", sheet_name=None)

plt.figure(figsize=(10,5))

for sheet_name, df in sheets.items():

    # Clean column names
    df.columns = df.columns.str.strip().str.lower()

    # Auto-detect date column
    date_col = [c for c in df.columns if 'date' in c or 'time' in c][0]

    # Convert to datetime
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

    # Pick first numeric column
    value_col = df.select_dtypes(include='number').columns[0]

    # ---- MONTHLY AGGREGATION ----
    df['month'] = df[date_col].dt.to_period('M')
    monthly = df.groupby('month')[value_col].mean()   # or .sum()

    # Convert Period → Timestamp for plotting
    monthly.index = monthly.index.to_timestamp()

    plt.plot(monthly.index, monthly.values, label=sheet_name)

plt.xlabel("Month")
plt.ylabel("CPI")
plt.title("Month-wise Time Series")
plt.legend()
plt.grid(True)
plt.show()

```

```python
import pandas as pd

# =====================================================
# 1. FILE PATHS  (CHANGE THESE)
# =====================================================
file1 = "IPI_94.xlsx"
file2 = "IP_5.xlsx"
file3 = "IPI_11.xlsx"

# =====================================================
# 2. FUNCTION TO CLEAN EACH FILE
# Assumes:
#   Column 1 = Date (DD-MM-YYYY)
#   Column 2 = IPI
# =====================================================
def load_ipi(file):
    df = pd.read_excel(file).iloc[:, :2]
    df.columns = ["Date", "IPI"]
    df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")
    df = df.dropna()
    df = df.set_index("Date").sort_index()
    return df

s1 = load_ipi(file1)
s2 = load_ipi(file2)
s3 = load_ipi(file3)

# =====================================================
# 3. STEP A — LINK FILE2 → FILE3 USING JAN 2013
# =====================================================
link_date = pd.Timestamp("2013-01-01")

R23 = s3.loc[link_date, "IPI"] / s2.loc[link_date, "IPI"]
s2_adj = s2 * R23

# =====================================================
# 4. STEP B — LINK FILE1 → ADJUSTED FILE2
# Use last overlapping date between s1 and s2
# =====================================================
overlap_date = s1.index.intersection(s2_adj.index).max()

R12 = s2_adj.loc[overlap_date, "IPI"] / s1.loc[overlap_date, "IPI"]
s1_adj = s1 * R12

# =====================================================
# 5. STEP C — CREATE FINAL CONTINUOUS SERIES
# =====================================================
final_ipi = pd.concat([
    s1_adj[s1_adj.index < s2_adj.index.min()],
    s2_adj[(s2_adj.index >= s1_adj.index.max()) & (s2_adj.index < s3.index.min())],
    s3
]).sort_index()

# =====================================================
# 6. SAVE OUTPUT
# =====================================================
final_ipi.to_excel("spliced_ipi.xlsx")

print("✅ Splicing completed successfully.")
print("📁 Output saved as: spliced_ipi.xlsx")

# =====================================================
# 7. PLOT THE SPLICED IPI
# =====================================================
plt.figure(figsize=(10, 5))
plt.plot(final_ipi.index, final_ipi["IPI"])
plt.title("Spliced Industrial Production Index (IPI)")
plt.xlabel("Date")
plt.ylabel("IPI (2011–12 Base Linked)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================
# 1. FILE PATHS
# =====================================================
files = {
    "IPI": "spliced_ipi.xlsx",
    "CPI": "Key Growth Rate YoY Consumer Price Index.xlsx",
    "GPR": "data_gpr_export.xls",
    "RATE": "Call money rate.xlsx",
    "UNCERT": "US_Policy_Uncertainty_Data.xlsx",
    "MPU": "US_MPU_Monthly.xlsx",
    "TPU": "tpu_web_latest.xlsx",
    "USVIX": "VIXCLS.xlsx",
    "FX": "FX USD.xlsx",
    "INVIX": "NSE VIX india.xlsx",
    "OUTPUTGAP": "ipi_with_output_gap.xlsx"
}

# =====================================================
# 2. MIXED DATE PARSER
# =====================================================
def parse_mixed_dates(date_series):
    dates = pd.to_datetime(date_series, errors="coerce", dayfirst=True)

    mask = dates.isna()
    if mask.any():
        dates.loc[mask] = pd.to_datetime(
            date_series[mask],
            format="%m-%Y",
            errors="coerce"
        )
    return dates


# =====================================================
# 3. LOAD EACH SERIES (STRICTLY FIRST 2 COLUMNS)
# =====================================================
def load_series(file, col_name):

    # ---- TAKE ONLY FIRST TWO COLUMNS ----
    df = pd.read_excel(file).iloc[:, :2].copy()
    df.columns = ["Date", col_name]

    # ---- PARSE DATES ----
    df["Date"] = parse_mixed_dates(df["Date"])

    # ---- DROP INVALID ROWS ----
    df = df.dropna(subset=["Date", col_name])

    # ---- CONVERT TO MONTHLY ----
    df["Month"] = df["Date"].dt.to_period("M")
    df = df.groupby("Month")[col_name].mean()

    # ---- TIMESTAMP INDEX ----
    df.index = df.index.to_timestamp()

    # ---- ENSURE UNIQUE INDEX ----
    df = df[~df.index.duplicated(keep="first")].sort_index()

    return df.to_frame()


# =====================================================
# 4. LOAD ALL DATA
# =====================================================
data_frames = []

for name, path in files.items():
    try:
        df = load_series(path, name)
        data_frames.append(df)
        print(f"Loaded: {name}")
    except Exception as e:
        print(f"Skipped {name}: {e}")


# =====================================================
# 5. MERGE INTO FINAL DATASET
# =====================================================
final_df = pd.concat(data_frames, axis=1)

# Keep common sample only (important for correlation)
final_df = final_df.dropna()

# Save merged dataset
final_df.to_excel("final_macro_dataset.xlsx")
print("Merged dataset saved as final_macro_dataset.xlsx")


# =====================================================
# 6. HEATMAP USING ONLY NUMERIC COLUMNS
# (guarantees only variable columns used)
# =====================================================
numeric_df = final_df.select_dtypes(include="number")

corr_matrix = numeric_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Heatmap of Macroeconomic Variables")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================
# 1. FILE PATHS
# =====================================================
files = {
    # "IPI": "spliced_ipi.xlsx",
    # "CPI": "Key Growth Rate YoY Consumer Price Index.xlsx",
    "GPR": "data_gpr_export.xls",
    # "RATE": "Call money rate.xlsx",
    "UNCERT": "US_Policy_Uncertainty_Data.xlsx",
    "MPU": "US_MPU_Monthly.xlsx",
    #"TPU": "tpu_web_latest.xlsx",
    # "USVIX": "VIXCLS.xlsx",
    # "FX": "FX USD.xlsx",
    # "INVIX": "NSE VIX india.xlsx",
    # "OUTPUTGAP": "ipi_with_output_gap.xlsx"
}

# =====================================================
# 2. MIXED DATE PARSER
# =====================================================
def parse_mixed_dates(date_series):
    dates = pd.to_datetime(date_series, errors="coerce", dayfirst=True)

    mask = dates.isna()
    if mask.any():
        dates.loc[mask] = pd.to_datetime(
            date_series[mask],
            format="%m-%Y",
            errors="coerce"
        )
    return dates


# =====================================================
# 3. LOAD EACH SERIES (STRICTLY FIRST 2 COLUMNS)
# =====================================================
def load_series(file, col_name):

    # ---- Read Excel ----
    df = pd.read_excel(file)

    # ---- If only ONE column exists (like TPU) ----
    if df.shape[1] == 1:
        df = df.reset_index()  # assume index holds dates

    # ---- Take only first TWO columns safely ----
    df = df.iloc[:, :2].copy()

    # If still only one column → skip
    if df.shape[1] < 2:
        raise ValueError("File does not contain Date + Value columns")

    df.columns = ["Date", col_name]

    # ---- Parse mixed date formats ----
    df["Date"] = parse_mixed_dates(df["Date"])

    # ---- Drop invalid rows ----
    df = df.dropna(subset=["Date", col_name])

    # ---- Convert to MONTHLY ----
    df["Month"] = df["Date"].dt.to_period("M")
    df = df.groupby("Month")[col_name].mean()

    # ---- Timestamp index ----
    df.index = df.index.to_timestamp()

    # ---- Ensure unique index ----
    df = df[~df.index.duplicated(keep="first")].sort_index()

    return df.to_frame()



# =====================================================
# 4. LOAD ALL DATA
# =====================================================
data_frames = []

for name, path in files.items():
    try:
        df = load_series(path, name)
        data_frames.append(df)
        print(f"Loaded: {name}")
    except Exception as e:
        print(f"Skipped {name}: {e}")


# =====================================================
# 5. MERGE INTO FINAL DATASET
# =====================================================
final_df = pd.concat(data_frames, axis=1)

# Keep common sample only (important for correlation)
final_df = final_df.dropna()

# Save merged dataset
final_df.to_excel("final_macro_dataset.xlsx")
print("Merged dataset saved as final_macro_dataset.xlsx")


# =====================================================
# CLEAN FOR OVERLAP
# =====================================================
final_df = final_df.dropna(thresh=2)

# =====================================================
# CORRELATION WITH MINIMUM OVERLAP
# =====================================================
corr_matrix = final_df.corr(min_periods=12)

# Safety check
if corr_matrix.isna().all().all():
    raise ValueError("No overlapping time period across variables.")

# =====================================================
# HEATMAP
# =====================================================
plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Correlation Heatmap of Macroeconomic Variables")
plt.tight_layout()
plt.show()

```
